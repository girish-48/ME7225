```python
import numpy.random as rand
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
values_100 = np.sort(rand.normal(size=100))
values_1000 = np.sort(rand.normal(size=1000))
```

```python
plt.plot(values_100, norm.pdf(values_100),color='red')
plt.plot(values_1000, norm.pdf(values_1000),color='black')
plt.show()
```

```python
generated_values = np.sort(rand.randn(150000))

mean = 10
std  = 2

y_gen_vals = std*generated_values + mean
# y          = np.sort(rand.normal(loc=10,scale=2,size=150000))
yvec       = np.linspace(0,20,1000)

# plt.hist(y, bins=100, density=True, alpha=0.75, color='r')
plt.hist(y_gen_vals, bins=100, density=True, alpha=0.65, color='b')
plt.plot(yvec, (1/(2*np.pi*std*std)**0.5)*np.exp(-(yvec - mean)**2/(2*std*std)), color='black')
plt.show()
```

```python
x = rand.randn(200000)
y = rand.randn(200000)

z = np.sort((x+y)/2)
zvec = np.linspace(-5,5,1000)

plt.hist(z,bins=500,density=True)
plt.plot(zvec, (1/(2*np.pi*0.5)**0.5)*np.exp(-zvec**2))
plt.show()
```

```python
x1 = rand.normal(loc=mean, scale=std, size=200000)
x2 = rand.normal(loc=mean, scale=std, size=200000)
x3 = rand.normal(loc=mean, scale=std, size=200000)
x4 = rand.normal(loc=mean, scale=std, size=200000)

z = (x1+x2+x3+x4)/4
zvec = np.linspace(-5,25,1000)

std1 = std/2

plt.hist(z,bins=70,density=True)
plt.plot(zvec, (1/(2*np.pi*std1*std1)**0.5)*np.exp(-(zvec-mean)**2/(2*std1*std1)))
plt.show()
```
