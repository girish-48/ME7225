## Fast Fourier Transform 
* Decimation in time 
$$ X[k] = E[k] + W_{N}^{k}\cdot O[k]

```python
def W(N) :
    return np.exp(-1.0j*2*np.pi/N)
```

```python
# Works for any one dimensional array as input but will not work for 2d arrays
# Make sure that the input is one dimensional before passing 

def dit(x,k):
    
    N = len(x)
    
    if N == 1:
        return x
    else:
        X_even = dit(x[::2],k)
        X_odd = dit(x[1::2],k)
        factor = np.exp(-2j*np.pi*k/N)
        
        return X_even + X_odd*factor
```

```python
def dit2(x):
    
    N = len(x)
    
    if N == 1:
        return x
    else:
        E = dit2(x[::2])
        O = dit2(x[1::2])
        factor = np.exp(-2j*np.pi*np.arange(N)/N)

         
        X = np.concatenate([E+factor[:int(N/2)]*O, E+factor[int(N/2):]*O])
        
        return X
```

### DFT Matrix Computation

```python
DFT = np.empty((N,N),dtype=complex)

for i in range(0,N) :  
    for j in range(0,N) : 
        DFT[i,j] = W(N)**(i*j)
```
