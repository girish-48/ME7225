```python
import numpy as np
import pandas as pd

# load the data from the csv file
data = pd.read_csv('qset3.csv').to_numpy()

# extract the input and output samples for n >= 1
x = data[1:, 0]
y = data[1:, 1]

# construct the matrix X
N = len(y)
X = np.zeros((N, 2))
for n in range(N):
    X[n, :] = [x[n], x[n-1] if n > 0 else 0]

# solve the least squares problem
p_hat = np.linalg.inv(X.T @ X) @ X.T @ y

# estimate a and b
a_hat = p_hat[0]
b_hat = p_hat[1]

# print the estimates
print('a_hat = {:.4f}'.format(a_hat))
print('b_hat = {:.4f}'.format(b_hat))

```

```python
import numpy as np

# x = data['input']
d = y

# Number of filter taps
N = 2

# Length of input signal
M = len(x)

# Calculate sample autocorrelation function
r = np.correlate(x, x, mode='full') / M

# Construct autocorrelation matrix
R = np.zeros((N, N))
for i in range(N):
    for j in range(N):
        R[i,j] = r[np.abs(i-j)]

# Calculate inverse of autocorrelation matrix
R_inv = np.linalg.inv(R)

# Calculate cross-correlation vector
p = np.zeros((N, 1))
for i in range(N):
    p[i] = np.sum(d * np.roll(x[::-1], i))

# Calculate filter tap estimate vector
h_hat = np.dot(R_inv, p)

# Calculate estimated output signal
y = np.convolve(x, h_hat, mode='valid')

# Calculate error vector
e = d - y

# Calculate variance of estimate for h[0]
var_h0 = R_inv[0,0] * 1

# Calculate variance of estimate for h[1]
var_h1 = R_inv[1,1] * 1

# Calculate covariance between estimate of h[0] and h[1]
cov_h0_h1 = -R_inv[0,1] * 1

# Print results
print("Variance of estimate for h[0]: {:.4f}".format(var_h0))
print("Variance of estimate for h[1]: {:.4f}".format(var_h1))
print("Covariance between estimate of h[0] and h[1]: {:.4f}".format(cov_h0_h1))

```

```python
import numpy as np

# Define input sequence x
N = 100
n = np.arange(N)
x = np.sin(2 * np.pi * n / N) + 0.5 * np.sin(4 * np.pi * n / N)

# Compute DFT of x
X = np.fft.fft(x)

# Compute DTFT at frequencies k1 and k2
k1 = 0.2
k2 = 0.3
exp_seq1 = np.exp(-1j * 2 * np.pi * k1 * n / N)
exp_seq2 = np.exp(-1j * 2 * np.pi * k2 * n / N)
dtft1 = np.sum(X * exp_seq1)
dtft2 = np.sum(X * exp_seq2)

# Print results
print(f"DTFT at frequency k={k1} is {dtft1}")
print(f"DTFT at frequency k={k2} is {dtft2}")
```
