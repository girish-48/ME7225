```python
import numpy as np 
import matplotlib.pyplot as plt
from scipy.io import wavfile
```

```python
samplerate, data  = wavfile.read('pure_tone.wav')
```

```python
samplerate
```

```python
data.shape
```

```python
data.shape[0]/samplerate
```

```python
dt = 1/samplerate
nvec = np.arange(0, data.shape[0])
tvec = dt*nvec
plt.plot(tvec,nvec)
```

```python
T = tvec[-1] + dt
df = 1/T
freq_vec = df*nvec
data_freq = np.fft.fft(data)
data_freq2 = np.fft.fftshift(data_freq)
N = data.shape[0]
nvec2 = np.arange(-N/2, N/2, 1)
freq_vec2 = df*nvec2

%matplotlib notebook

plt.stem(freq_vec2, np.abs(data_freq2), use_line_collection=True)
```
