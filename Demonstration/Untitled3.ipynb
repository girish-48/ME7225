# FOURIER SERIES PROPERTIES 

```python
# importing all necessary python modules 

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt 
from scipy import integrate
```

```python
# variables and mathemetical function definition

T = 2
w0 = 2*np.pi/T

t = sp.symbols("t")

# All the signals are defined from -2*T to 2*T and have been used piecewise values without modulus to obtian the result

square     = sp.Piecewise( (1,(t>-2*T)&(t<= -3*T/2)), (0, (t<= -T)&(t>-3*T/2)), (1,(t>-T)&(t<= -T/2)), (0, (t<=0)&(t>-T/2)), (0, (t>=T/2)&(t<=T)), (1, (t>0)&(t<T/2)), (0, (t>=3*T/2)&(t<2*T)), (1, (t>T)&(t<=3*T/2)) )
triangular = sp.Piecewise( (2*T+t, (t>-2*T)&(t<= -3*T/2)), (-t-T, (t>-3*T/2)&(t<= -T/2)), (t,(t>-T/2)&(t<=T/2)), (T - t, (t>T/2)&(t<=3*T/2)), (t - 2*T, (t>3*T/2)&(t<=2*T)) )
sawtooth   = sp.Piecewise( (t+2*T, (t>-2*T)&(t<= -3*T/2)), (t+T, (t>-3*T/2)&(t<= -T/2)), (t, (t>-T/2)&(t<=T/2)), (t-T, (t>T/2)&(t<=3*T/2)), (t-2*T, (t>3*T/2)&(t<2*T)))
```

```python
# Plotting the signal

tvec = np.linspace(-2*T, 2*T, 1000, endpoint=False)
xnum = sp.lambdify(t, square*sp.exp(1.0j*t))
plt.plot(tvec, np.imag(xnum(tvec)))
```
