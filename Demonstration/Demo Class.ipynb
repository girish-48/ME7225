Basic features of symbolic and numerical computation using Jupyter/ Python

```python
import numpy as np
# numpy is the module associated with the data structures associated with numerical computation
import sympy as sp 
# sympy is the module associated with symbolic computation 
import matplotlib.pyplot as plt 
# this is used for plotting 
from scipy import integrate
# numerical integration
```

## Symbolic Computation 

Define the symbolic variables 

```python
t, alpha = sp.symbols('t, alpha')
```

```python
t
```

```python
alpha
```

```python
x = sp.sin(t)
```

```python
dx = sp.diff(x)
```

```python
dx
```

```python
sp.N(sp.integrate(x, (t, 0, 1)), n=20)
```

```python
tnum = 1.0 
np.sin(tnum)
```

```python
x.subs(t, 1.0)
```

## Plotting

- plotting can be done only for numerical values 

```python
tvec = np.linspace(0, 10, 100, endpoint= False)
```

```python
plt.plot(tvec, np.sin(tvec))
plt.plot(tvec, np.cos(tvec), 'r--')
plt.grid()
plt.xlabel('Time')
plt.title('Demonstration')
```
